````python
from IPython.display import Markdown, display
import json

def md_show(response):
    """Display API response in a formatted Markdown way."""
    status = response.status_code
    try:
        body = json.dumps(response.json(), indent=4)
    except:
        body = response.text
    
    md_text = f"""
### API Response  
**Status:** `{status}`  

```json
{body}
```"""
    display(Markdown(md_text))
````

## 3. System Login

````python
USERNAME = "oderogeorge308@gmail.com" 
PASSWORD = "geo@123!" 
````

````python
def login(username: str, password: str):
    token_url = f"{BASE_URL}/auth/token"

    data = {
        "username": username,
        "password": password
    }

    headers = {
        "Content-Type": "application/x-www-form-urlencoded"
    }

    response = requests.post(token_url, data=data, headers=headers)

    if response.status_code == 200:
        token = response.json().get("access_token")
        # display(Markdown(f"### Login successful\n**Access Token:** `{token}`"))
        return token
    else:
        display(Markdown(f"### Login failed\n**Status:** {response.status_code}\n\n```\n{response.text}\n```"))
        return None
````

````python
def api_post(path: str, payload: dict):
    ACCESS_TOKEN = login(USERNAME, PASSWORD);
    if ACCESS_TOKEN is None:
        display(Markdown("You must login first."))
        return
    
    url = f"{BASE_URL}{path}"
    headers = {
        "Authorization": f"Bearer {ACCESS_TOKEN}",
        "Content-Type": "application/json"
    }
    
    response = requests.post(url, headers=headers, json=payload)

    # Format response body
    try:
        json_body = response.json()
        formatted = json.dumps(json_body, indent=4)
    except:
        formatted = response.text

    display(Markdown(f"""
### **Status:** `{response.status_code}`  

#### Request Body
```json
{json.dumps(payload, indent=4)}
{formatted}"""))
    return response
````

````python
def api_post_no_payload(path: str):
    ACCESS_TOKEN = login(USERNAME, PASSWORD);
    if ACCESS_TOKEN is None:
        display(Markdown("You must login first."))
        return
    
    url = f"{BASE_URL}{path}"
    
    headers = {
        "Authorization": f"Bearer {ACCESS_TOKEN}",
        "Content-Type": "application/json"
    }

    response = requests.post(url, headers=headers)

    # Try JSON, fallback to text
    try:
        formatted = json.dumps(response.json(), indent=4)
    except Exception:
        formatted = response.text
    
    display(Markdown(f"""
### **Status:** `{response.status_code}`  
```json
{formatted}
"""))
````

````python
# Helper function for calling protected API Endpoints
def api_get(path: str):
    ACCESS_TOKEN = login(USERNAME, PASSWORD);
    if ACCESS_TOKEN is None:
        display(Markdown("You must login first."))
        return

    url = f"{BASE_URL}{path}"

    headers = {
        "Authorization": f"Bearer {ACCESS_TOKEN}"
    }

    response = requests.get(url, headers=headers)

    try:
        json_body = response.json()
        formatted = json.dumps(json_body, indent=4)
    except:
        formatted = response.text

    display(Markdown(f"""
### **Status:** `{response.status_code}`  
```json
{formatted}
"""))
````

````python
from IPython.display import Markdown, display
import json

def display_emergency_analysis(api_response: dict):
    """
    Render the emergency analysis API response in a clean UI-friendly Markdown format.
    """

    # Extract fields safely
    response_text = api_response.get("response", "")
    reasoning = api_response.get("reasoning", "")
    actions = api_response.get("actions_taken", [])
    confidence = api_response.get("confidence", "")

    md = f"""
# Emergency Situation Analysis

## Summary
{response_text}

---

## Reasoning
{reasoning}

---

## Actions Taken
"""    
    for action in actions:
        md += f"- {action}\n"

    md += f"""

---

## Confidence Score  
**{confidence}**

"""

    display(Markdown(md))
````
