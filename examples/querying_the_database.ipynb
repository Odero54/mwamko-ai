```python
import geopandas as gpd
from sqlalchemy import create_engine

# 1. Connect to the database
cloud_engine = create_engine(
    "postgresql://postgres:postgres@127.0.0.1:5432/mwamko"
)

# 2. Function to load table from PostGIS as GeoDataFrame
def load_postgis_table(table_name, geom_col="geom"):
    """
    Load a PostGIS table as GeoDataFrame.
    """
    gdf = gpd.read_postgis(
        f"SELECT * FROM {table_name};",
        cloud_engine,
        geom_col=geom_col
    )
    return gdf

# 3. Load each table
constraints_gdf = load_postgis_table("constraints")
road_segments_gdf = load_postgis_table("road_segments")
vertices_gdf = load_postgis_table("road_segments_vertices_pgr")

# 4. Export to GeoJSON
constraints_gdf.to_file("/Users/georgeodero/mwamko-ai/datasets/processed/constraints.geojson", driver="GeoJSON")
road_segments_gdf.to_file("/Users/georgeodero/mwamko-ai/datasets/processed/road_segments.geojson", driver="GeoJSON")
vertices_gdf.to_file("/Users/georgeodero/mwamko-ai/datasets/processed/road_segments_vertices_pgr.geojson", driver="GeoJSON")

constraints_gdf.head()
```
